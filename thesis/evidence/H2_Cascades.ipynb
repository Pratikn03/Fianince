Registered in `thesis/checks/prereg/H2_cascade.json` before any of its data were analysed. It uses the Hyperliquid trades that the **H1 collector is already recording** in `MyDrive/h1_data` (nothing new to collect) and opens with H1: on or after **10 December 2026**, once at least 20 coins have 30 valid days. Before then cell 3 refuses to run.

* **P1** the feedback part of the stability margin, certified from public prices and anonymous flow, rises in the six hours before crash hours (each coin's top 1% of hourly excursions) more than before calm hours with the same prior volatility;
* **P2** the margin's decline forecasts the size of the next hour's excursion beyond a volatility baseline, out of sample.

**Runtime:** CPU is enough (no GPU). About 10–30 minutes. Run it in December, after H1's analysis, in the same Drive folder.


```python
# 1. Registered code and the H1 data folder on Drive
import os, subprocess, sys, json, shutil, zipfile
from pathlib import Path
REPO = "/content/Fianince"
if not os.path.exists(REPO):
    subprocess.run(["git", "clone", "-q", "https://github.com/Pratikn03/Fianince.git", REPO], check=True)
subprocess.run(["git", "-C", REPO, "pull", "-q", "origin", "main"], check=True)
sys.path.insert(0, f"{REPO}/thesis/evidence")
from google.colab import drive
drive.mount("/content/drive", timeout_ms=300000)
OUT = "/content/drive/MyDrive/h1_data"
assert (Path(OUT) / "H1_universe.json").exists(), "H1's data folder not found on Drive"
print("data:", OUT)
```

```python
# 2. Check the registration and whether the data gate is open (counts valid days only; does not analyse)
import h2_cascade as H2
REPO_COUNTER = Path(REPO) / "thesis/checks/prereg/H2_unlock.json"; DRIVE_COUNTER = Path(OUT) / REPO_COUNTER.name
if not DRIVE_COUNTER.exists():
    shutil.copy(REPO_COUNTER, DRIVE_COUNTER)
print("counter:", json.loads(DRIVE_COUNTER.read_text()))
spec, sha, commit = H2.gate()
print("specification sha256:", sha, "| registered in commit:", commit)
print("data gate:", H2.H.data_gate(OUT))
```
