# Test R1 — does the absolute response measure risk where the margin did not?

Registered in `thesis/checks/prereg/R1_absresp.json` **before** any of these data were downloaded. Run it **once**, **after S1's result has been committed** (cell 2 refuses until then, so that S1 and R1 cannot draw the same markets). About 2–3 hours.

**What it does.** It draws 40 Binance perpetuals that appear nowhere in the repository (seed 20261109), writes them down, then downloads every day from January 2025 to September 2026. Three registered primaries (Holm-corrected):

* **P1** within a market, the next day's largest adverse move rises with the estimated absolute response Ψ_H (instrumented by its lag, controlling for shock size);
* **P2** Ψ_H forecasts the next day's largest adverse move out of sample, beyond volatility;
* **P3** it does so beyond the margin.
