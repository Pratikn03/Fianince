# Test F1 — the flagship test: certify, don't identify, and does the certificate forecast risk?

Registered in `thesis/checks/prereg/F1_flagship.json` **before** any of these data were downloaded.
This notebook runs the test **once**. It needs about 2–3 hours in Colab (about 26,000 small downloads, then the analysis).

**What it does.** It draws 30 Binance USD-M perpetuals that appear nowhere in the repository (seed 20261105), writes that list down, and only then downloads 430 dates. Three registered primaries (Holm-corrected):

* **P1** the safety margin is decided in ≥ 75% of symbols;
* **P2** "decided but not identified" beats "identified but not decided" (McNemar);
* **P3** the margin from the previous 20 days forecasts the next day's largest adverse move out of sample, beyond volatility (pooled model, Clark–West test clustered by date, negative coefficient).

**Rules.** Run the cells in order. Do not edit any file in the repository clone. If the runtime disconnects during cell 3, run cells 1–2 again and set `RESUME = True` in cell 3. Whatever the result, save it (cell 5).

```python
# 1. Get the registered code and a Drive cache (the Drive mount is retried; a Google sign-in window will appear:
#    choose your account and click Allow on every screen)
import os, subprocess, sys, time
REPO = "/content/Fianince"
if not os.path.exists(REPO):
    subprocess.run(["git", "clone", "-q", "https://github.com/Pratikn03/Fianince.git", REPO], check=True)
subprocess.run(["git", "-C", REPO, "pull", "-q", "origin", "main"], check=True)
sys.path.insert(0, f"{REPO}/thesis/evidence")
from google.colab import drive
for attempt in range(3):
    try:
        drive.mount("/content/drive", force_remount=True, timeout_ms=300000)
        break
    except ValueError as e:
        print("mount attempt", attempt + 1, "failed:", e); time.sleep(10)
else:
    raise SystemExit("Drive did not mount: do not run the test; tell Claude")
CACHE = "/content/drive/MyDrive/f1_cache"
os.makedirs(CACHE, exist_ok=True)
print("Drive OK:", CACHE)
print(subprocess.run(["git", "-C", REPO, "log", "-3", "--format=%h %ci %s"], capture_output=True, text=True).stdout)
```

```python
# 2. Check the registration: specification hash, pipeline hashes, spec committed, counter at 0.
# The counter and the drawn universe are kept on Drive, so a disconnect can never reopen the test or redraw symbols.
import json, shutil
from pathlib import Path
assert "CACHE" in globals() and CACHE.startswith("/content/drive"), "run cell 1 first and wait for Drive OK"
import f1_flagship as C
REPO_COUNTER, REPO_UNIVERSE = Path(C.COUNTER), Path(C.UNIVERSE)
DRIVE_COUNTER, DRIVE_UNIVERSE = Path(CACHE) / REPO_COUNTER.name, Path(CACHE) / REPO_UNIVERSE.name
if not DRIVE_COUNTER.exists():
    shutil.copy(REPO_COUNTER, DRIVE_COUNTER)
C.COUNTER, C.UNIVERSE = DRIVE_COUNTER, DRIVE_UNIVERSE
print("counter:", json.loads(DRIVE_COUNTER.read_text()))
spec, sha, commit = C.gate()
print("specification sha256:", sha); print("registered in commit:", commit)

# Addendum A1 (registered before any data): universe symbols must also be ASCII. Verify its hash, then apply it.
import hashlib, f1_addendum
A1 = json.loads((Path(REPO) / "thesis/checks/prereg/F1_addendum_A1.json").read_text())
assert hashlib.sha256((Path(REPO) / "thesis/evidence/f1_addendum.py").read_bytes()).hexdigest() == A1["f1_addendum_sha256"]
assert A1["f1_spec_sha256"] == sha, "addendum does not refer to this specification"
f1_addendum.apply(); print("addendum A1 applied (ASCII symbols only)")
import c3_margin_more_days as C3
print(len(C3.dates()), "dates from", C3.dates()[0], "to", C3.dates()[-1])
```

```python
# 3. RUN THE TEST (once). The counter becomes 1, the universe is drawn and saved, then data are downloaded and evaluated.
# Only if the runtime disconnected after this cell had started: set RESUME = True (counter on Drive already 1;
# the saved universe on Drive is reused, so the same symbols are evaluated).
RESUME = True    # the test is already open (counter 1); keep True
res = C.run(cache_dir=CACHE, threads=16, resume=RESUME)
```

```python
# 4. Results
import pandas as pd
print(json.dumps(res["evaluation"], indent=1, default=float))
pd.DataFrame([{"symbol": s, "valid_days": v["n_days"], "margin": v.get("sign"), "identified": v.get("identified"),
               "note": v.get("excluded")} for s, v in res["per_symbol"].items()])
```
